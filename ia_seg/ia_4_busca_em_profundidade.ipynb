```python
import numpy as np
import pandas as pd

class Cidade:
  def __init__(self, nome):
    self.nome = nome
    self.visitado = False # Alterado de 'vizinhos' para 'visitado' e inicializado como False
    self.adjacente = []

  def addCidadeAdjacente(self, cidade):
    self.adjacente.append(cidade)
```

```python
class Adjacente:
  def __init__(self, cidade):
    self.cidade = cidade
```

```python
class Mapa:
  portoAlegre = Cidade('Porto Alegre')
  canoas = Cidade('Canoas')
  eldoradoDoSul = Cidade('Eldorado do Sul')
  venancioAires = Cidade("Venancio Aires")
  santaCruz = Cidade("Santa Cruz")
  candelaria = Cidade("Candelaria")
  santaMaria = Cidade("Santa Maria")
  agudo = Cidade("Agudo")
  restingaSeca = Cidade("Restinga Seca")
  tupancy = Cidade("Tupancy")
  rioPardo = Cidade("Rio Pardo")
  pantanoGrande = Cidade("Pantano Grande")
  saoSepe = Cidade("São Sepé")
  triunfo = Cidade("Triunfo")
  cachoeiraDoSul = Cidade("Cachoeira do Sul")
  guaiba = Cidade("Guaíba")

  portoAlegre.addCidadeAdjacente(Adjacente(canoas))
  portoAlegre.addCidadeAdjacente(Adjacente(eldoradoDoSul))
  portoAlegre.addCidadeAdjacente(Adjacente(cachoeiraDoSul))

  canoas.addCidadeAdjacente(Adjacente(portoAlegre))
  canoas.addCidadeAdjacente(Adjacente(venancioAires))

  eldoradoDoSul.addCidadeAdjacente(Adjacente(portoAlegre))
  eldoradoDoSul.addCidadeAdjacente(Adjacente(guaiba))
  eldoradoDoSul.addCidadeAdjacente(Adjacente(pantanoGrande))

  venancioAires.addCidadeAdjacente(Adjacente(canoas))
  venancioAires.addCidadeAdjacente(Adjacente(santaCruz))
  venancioAires.addCidadeAdjacente(Adjacente(cachoeiraDoSul))

  santaCruz.addCidadeAdjacente(Adjacente(venancioAires))
  santaCruz.addCidadeAdjacente(Adjacente(cachoeiraDoSul))
  santaCruz.addCidadeAdjacente(Adjacente(candelaria))

  candelaria.addCidadeAdjacente(Adjacente(santaCruz))
  candelaria.addCidadeAdjacente(Adjacente(agudo))
  candelaria.addCidadeAdjacente(Adjacente(santaMaria))

  santaMaria.addCidadeAdjacente(Adjacente(candelaria))
  santaMaria.addCidadeAdjacente(Adjacente(agudo))
  santaMaria.addCidadeAdjacente(Adjacente(restingaSeca))
  santaMaria.addCidadeAdjacente(Adjacente(tupancy))

  agudo.addCidadeAdjacente(Adjacente(santaMaria))
  agudo.addCidadeAdjacente(Adjacente(candelaria))
  agudo.addCidadeAdjacente(Adjacente(rioPardo))

  restingaSeca.addCidadeAdjacente(Adjacente(santaMaria))
  restingaSeca.addCidadeAdjacente(Adjacente(rioPardo))

  tupancy.addCidadeAdjacente(Adjacente(santaMaria))
  tupancy.addCidadeAdjacente(Adjacente(saoSepe))

  rioPardo.addCidadeAdjacente(Adjacente(agudo))
  rioPardo.addCidadeAdjacente(Adjacente(restingaSeca))
  rioPardo.addCidadeAdjacente(Adjacente(triunfo))

  pantanoGrande.addCidadeAdjacente(Adjacente(saoSepe))
  pantanoGrande.addCidadeAdjacente(Adjacente(rioPardo))
  pantanoGrande.addCidadeAdjacente(Adjacente(eldoradoDoSul))

  saoSepe.addCidadeAdjacente(Adjacente(pantanoGrande))
  saoSepe.addCidadeAdjacente(Adjacente(tupancy))

  triunfo.addCidadeAdjacente(Adjacente(rioPardo))
  triunfo.addCidadeAdjacente(Adjacente(cachoeiraDoSul))
  triunfo.addCidadeAdjacente(Adjacente(pantanoGrande))

  cachoeiraDoSul.addCidadeAdjacente(Adjacente(santaCruz))
  cachoeiraDoSul.addCidadeAdjacente(Adjacente(venancioAires))
  cachoeiraDoSul.addCidadeAdjacente(Adjacente(triunfo))
  cachoeiraDoSul.addCidadeAdjacente(Adjacente(guaiba))
  cachoeiraDoSul.addCidadeAdjacente(Adjacente(portoAlegre))

  guaiba.addCidadeAdjacente(Adjacente(cachoeiraDoSul))
  guaiba.addCidadeAdjacente(Adjacente(eldoradoDoSul))
```

```python

class Pilha:
  def __init__(self, tamanho):
    self.tamanho = tamanho
    self.cidades = [None] * self.tamanho
    self.topo = -1 #índice do último elemento válido( inicio vazio)

  def empilhar(self, cidade):
    if not Pilha.pilhaCheia(self):
      self.topo +=1
      self.cidades[self.topo] = cidade
    else:
      print("Pilha cheia")

  def desempilhar(self):
    if not Pilha.pilhaVazia(self):
      temp = self.cidades[self.topo]
      self.topo -=1
      return temp
    else:
      print("Pilha vazia")
      return None

  def getTopo(self):
    return self.cidades[self.topo]

  def pilhaVazia(self):
    return self.topo == -1

  def pilhaCheia(self):
    return self.topo == self.tamanho -1
```

```python
mapa = Mapa()
pilha = Pilha(5)
pilha.tamanho
pilha.topo
```

```python
pilha.empilhar(mapa.portoAlegre)
pilha.empilhar(mapa.canoas)
pilha.empilhar(mapa.eldoradoDoSul)
pilha.empilhar(mapa.venancioAires)
pilha.topo
pilha.getTopo().nome
pilha.desempilhar()
pilha.getTopo().nome
```

```python

class Profundidade:
  def __init__(self, inicio, objetivo):
    self.inicio = inicio
    self.inicio.visitado = True
    self.objetivo = objetivo
    self.fronteira = Pilha(20)
    self.fronteira.empilhar(inicio)

  def buscar(self):
    topo = self.fronteira.getTopo()
    print('Topo: {}'.format(topo.nome))

    # Corrigido de 'topo.adjacentes' para 'topo.adjacente'
    for a in topo.adjacente:
        print('Verificando se já foi visitado: {}'.format(a.cidade.nome))
        if a.cidade.visitado == False:
          a.cidade.visitado = True
          self.fronteira.empilhar(a.cidade)
          Profundidade.buscar(self)
    print('Desempilhou: {}'.format(self.fronteira.desempilhar().nome))
```

```python
mapa = Mapa()
profundidade = Profundidade(mapa.portoAlegre, mapa.canoas)
profundidade.buscar()
```
