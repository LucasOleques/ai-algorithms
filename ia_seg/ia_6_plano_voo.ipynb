```python
voos = {}

for linha in open('ia_6_flights.ipynb'):
    partes = linha.strip().split(',')

    if len(partes) == 5 and partes[4].isdigit():
        origem, destino, saida, chegada, preco = partes

        voos.setdefault((origem, destino), [])
        voos[(origem, destino)].append((saida, chegada, int(preco)))

        voos.setdefault((destino, origem), [])
        voos[(destino, origem)].append((chegada, saida, int(preco)))
    print(origem, destino, saida, chegada, preco)
```
