```python
# Estudo de caso ir de Arad a Bucharest usando o algoritmo
class Vertice:
  
  def __init__(self, rotulo, distancia_objeto):
    self.rotulo = rotulo
    self.visitado = False
    self.distancia_objeto = distancia_objeto
    self.adjacentes = []

  def adiciona_adjacente(self, adjacente):
    self.adjacentes.append(adjacente)

  def mostra_adjacentes(self):
    for i in self.adjacentes:
      print(i.vertice.rotulo, i.custo)
```

```python
class Adjacente:
    def __init__(self, vertice, custo):
        self.vertice = vertice
        self.custo = custo
```

```python
import numpy as np   # Importa a biblioteca NumPy para trabalhar com arrays

# Classe que implementa um vetor ordenado de vértices
class VetorOrdenado:

  # Construtor da classe
  def __init__(self, capacidade):
    self.capacidade = capacidade              # Tamanho máximo do vetor
    self.ultima_posicao = -1                  # Índice da última posição ocupada (-1 significa vazio)
    self.valores = np.empty(self.capacidade, dtype=object)
    # Cria um array vazio de objetos (nesse caso, vértices)

  # Método para inserir um vértice no vetor de forma ordenada
  def insere(self, vertice):
    # Verifica se já atingiu a capacidade máxima
    if self.ultima_posicao == self.capacidade - 1:
      print('Capacidade máxima atingida')
      return

    posicao = 0
    # Procura a posição correta onde o vértice deve ser inserido
    for i in range(self.ultima_posicao + 1):
      posicao = i
      # Critério de ordenação: distância para o objetivo
      if self.valores[i].distancia_objeto > vertice.distancia_objeto:
        break
      # Se chegou até o final sem achar posição, coloca no final
      if i == self.ultima_posicao:
        posicao = i + 1

    # Desloca os elementos à direita para abrir espaço para o novo vértice
    x = self.ultima_posicao
    while x >= posicao:
      self.valores[x + 1] = self.valores[x]
      x -= 1

    # Insere o novo vértice na posição correta
    self.valores[posicao] = vertice
    self.ultima_posicao += 1   # Atualiza a posição do último elemento

  # Método para imprimir os elementos do vetor
  def imprime(self):
    if self.ultima_posicao == -1:
      print('O vetor está vazio')
    else:
      # Exibe índice, rótulo e distância até o objetivo de cada vértice
      for i in range(self.ultima_posicao + 1):
        print(i, ' - ', self.valores[i].rotulo, ' - ', self.valores[i].distancia_objeto)
```

```python
class Gulosa:
  def __init__(self, objetivo):
    self.objetivo = objetivo
    self.encontrado = False

  def buscar(self, atual):
    print('-------')
    print('Atual: {}'.format(atual.rotulo))
    atual.visitado = True

    if atual == self.objetivo:
      self.encontrado = True
    else:
      vetor_ordenado = VetorOrdenado(len(atual.adjacentes))
      for adjacente in atual.adjacentes:
        if adjacente.vertice.visitado == False:
          adjacente.vertice.visitado == True
          vetor_ordenado.insere(adjacente.vertice)
      vetor_ordenado.imprime
```
